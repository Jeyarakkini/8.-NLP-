```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Embedding, Dense
```

```python
text = """
You are capable of more than you think. Every small step you take is proof of your strength.
You are learning, you are growing, and that is enough. You deserve good things, even on the days you doubt yourself.
Your effort matters, even when nobody notices. Progress may feel slow, but it is still progress.

It’s okay to feel confused; clarity comes after consistency. You don’t need to be perfect; you only need to try.
Your feelings are real, and they deserve respect. You are not falling behind; you are moving at your own pace.
You’ve overcome challenges before — you will overcome this too.

Every challenge is a chance to grow stronger. Mistakes are lessons in disguise; don’t fear them.
Curiosity is your greatest strength — keep asking questions. Improvement is a journey of patience and kindness to yourself.
Skills take time to build; confidence takes practice.

Sometimes you don’t know where you’re going, but you still need to move.
Some days are heavy, but you still choose to show up — that’s bravery.
It’s normal to want reassurance; it means you care.
You don’t have to be motivated every day; discipline will carry you.
You’re learning to trust yourself again, and that’s beautiful.

I am worthy of respect. I can learn anything with enough time.
I choose progress over perfection. I am strong, even when I feel weak.
I trust myself more every day.

There was a time when you doubted your abilities, but you still took the first step.
That moment changed everything. You once thought you couldn’t handle pressure,
but you adapted, survived, and grew. You proved to yourself that hard work beats talent
when talent doesn’t work hard. You’ve had failures, but every time, you stood up stronger.
Your journey is messy, painful, beautiful, and uniquely yours — and it’s shaping you.

Imagine waking up one day feeling proud because you didn’t quit.
Imagine someone telling you that your journey inspired them.
Imagine looking back and realizing this difficult phase was the turning point.
Imagine becoming the version of yourself you’re afraid to believe in.
"""
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

word_index = tokenizer.word_index
total_words = len(word_index) + 1

print("Total words:", total_words)
```

```python
input_sequences = []

for line in text.split("\n"):
    tokens = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(tokens)):
        seq = tokens[:i+1]
        input_sequences.append(seq)

print("Total sequences:", len(input_sequences))
```

```python
#Pad Sequences. Make all sequences equal length.
max_len = max([len(x) for x in input_sequences])
input_sequences = np.array(pad_sequences(input_sequences, maxlen=max_len, padding='pre'))

X = input_sequences[:, :-1]
y = input_sequences[:, -1]
```

```python
y
```

```python
# build lstm model
model = Sequential()
model.add(Embedding(total_words, 64, input_length=max_len-1))
model.add(LSTM(64))
model.add(Dense(total_words, activation='softmax'))

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.summary()
```

```python
y = tf.keras.utils.to_categorical(y, num_classes=total_words)
```

```python
y
```

```python
history = model.fit(X, y, epochs=200, verbose=1)
```

```python
history
```

```python
#predict next words
def predict_next_word(seed_text):
    token_list = tokenizer.texts_to_sequences([seed_text])[0]
    token_list = pad_sequences([token_list], maxlen=max_len-1, padding='pre')
    predicted = model.predict(token_list, verbose=0)
    predicted_index = np.argmax(predicted)
    
    for word, index in word_index.items():
        if index == predicted_index:
            return word
```

```python
print(predict_next_word("clarity comes after"))
```
